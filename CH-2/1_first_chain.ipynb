# **Chains in Langchain**

```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv


#This function will load all the variable from the .env file and will 
#make them available in the os.environ dictionary(env variables)
load_dotenv()

if os.environ.get("GROQ_API_KEY"):
    print("Bro, API KEY variable exists!")
else:
    raise ValueError("GROQ_API_KEY not found")

from langchain_core.messages import HumanMessage,SystemMessage,AIMessage
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser


llm_groq = ChatOpenAI(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1")
```

# **First Chain**

```python
#TASK1(prompt)
prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant"),
    ("human","{input}")
])
```

```python
#TASK2(LLM)

llm_groq = ChatOpenAI(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1")
```

```python
#TASK3-(Str parser)

str_parser = StrOutputParser()
```

# **Manual Invocation**

```python
template = prompt_template.invoke({"input":"what is the capital of France?"})

res = llm_groq.invoke(template)

final_result = res.content
```

# **Chain Invocation**

```python
chain = prompt_template | llm_groq | str_parser

chain.invoke({"input":"what is the capital of france?"})
```

OR

```python
from langchain_core.runnables import RunnableSequence

chain_1 = RunnableSequence(prompt_template,llm_groq,str_parser)

chain_1.invoke("what is the capital of france?")
```
