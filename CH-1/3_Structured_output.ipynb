# **Guiding in prompts**

```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv


#This function will load all the variable from the .env file and will 
#make them available in the os.environ dictionary(env variables)
load_dotenv()

if os.environ.get("GROQ_API_KEY"):
    print("Bro, API KEY variable exists!")
else:
    raise ValueError("GROQ_API_KEY not found")
```

```python
from langchain.chat_models import init_chat_model

llm_groq = init_chat_model(
    model="openai/gpt-oss-20b",
    model_provider="openai",
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)
```

```python
from langchain_core.prompts import PromptTemplate

result=llm_groq.invoke("Tell me a joke.Generate the output in key-value pair format with the following keys:setup,punchline")
result.content
```

# **Using Pydantic Models**

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup:str = Field(description="The setup for the joke")
    punchline:str=Field(description="The punchline for the joke")
```

```python
llm_structured_output=llm_groq.with_structured_output(llm_schema)

result=llm_structured_output.invoke("Tell me a joke")
result.punchline
```

# **Using TypeDict**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup:str
    punchline:str
```

```python
obj = llm_schema_td({"ketchup":"some setup","puchline":"some punchline"})
obj
```

```python
llm_structured_typed_dict=llm_groq.with_structured_output(llm_schema_td)

result = llm_structured_typed_dict.invoke("Tell me a joke")
result
```
