```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv


#This function will load all the variable from the .env file and will 
#make them available in the os.environ dictionary(env variables)
load_dotenv()

if os.environ.get("GROQ_API_KEY"):
    print("Bro, API KEY variable exists!")
else:
    print("API KEY not found!")
```

```python
import os
from openai import OpenAI


client = OpenAI(
    # This is the default and can be omitted
     api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)
response = client.responses.create(
    model="openai/gpt-oss-20b",
    instructions="You are a coding assistant that talks like a pirate.",
    input="How do I check if a Python object is an instance of a class?",
)
print(response.output_text)
```

```python
from langchain_openai import ChatOpenAI
import os 

llm_groq = ChatOpenAI(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1")

llm_groq.invoke("Bro,tell me a fun fact")
```

```python
from langchain.chat_models import init_chat_model

llm_groq = init_chat_model(model="openai/gpt-oss-20b",model_provider="openai", api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1")

llm_groq.invoke("Hello,how are you?")
```

```python
from langchain.chat_models import init_chat_model
# Example with Ollama
local_llm = init_chat_model(
   model="ollama:qwen2.5:3b",
   temperature=0,
   max_tokens=512
)
response = local_llm.invoke("Hello, how are you?")
print(response)
```
